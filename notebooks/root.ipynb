## Bracketing method  
- some terms: 
    - xl - lower bound x 
    - xu - upper bound x 
    - xr - current root guess 

- conditions:
    - a function
    - continuous between [a,b]
    - a sign change between f(a) and f(b) such that f(a)f(b)<0  
- Convergence 
    - as long as the initial guesses do bracket a root, these methods will always converge 
    - they might however take longer then an open method when converging 
- **Bisection**
    - brackets needed where it thinks a root exists
    - halves the interval and checks if a root exists there by preforming f(xr)*f(xl)   
- **False-Position**
    - adds weight to the xl and xu by evaluating if one value is closer to the root 
    - when modified, program evaluates if the same xl or xu has been used more then twice and halves that value in order to converge more rapidly 
- Preforms:     
    - if f(xr)*f(xl) < 0, the root is between f(xr) and f(xl), xu is now = xr 
    - if f(xr)*f(xl) > 0 the root is between f(xr) and f(xu), xl is now = xr 
    - if f(xr)*f(xl) = 0 the root is found 
    

## Open Methods 
- some terms:
    - x0: previous x  (only used for secant)
    - x: current x 
    - x1: next x value being solved for 
- Conditions: 
    - one or two initial x values that do not need to bracket a root 
- Convergence: 
    - Open methods can converge or diverge 
    - when converging, they tend to do so much quicker then bracketing 
- **Simple fixed point iteration** 
    - requires a function that is solvable for x (meaning bringing setting all of the equation equal to one of its variables) such that $x_{i+1}$ = g(x)
    - difficulties: 
        - can be difficult or even impossible to do in certain cases 
- **The secant method** 
    - requires two initial estimates that do not need to bracket 
    - benefits:
        - does not require a derivative function to be evaluated, instead approximates this through the use of a backward finite divided difference which is then subsituted into the iterative equation 
- **Newton-Raphson** 
    - requires the use of a derivative of the function in order to find the root
